```python
import numpy as np

def dit_fft(x):#define function
    N = len(x)#length of x
    if N <= 1:#if length of n is 0 or 1 return input
        return x#return function
    if N % 2 != 0: #check if power of 2
        raise ValueError("Size of x must be a power of 2")#raise an error

    # Divide: even and odd indices
    even = dit_fft(x[0::2])#x[0]+x[0+2]+x[2+2]+.....x[(N-2)+2]
    odd = dit_fft(x[1::2])#x[1]+x[0+3]+x[2+5]+.....x[(N-3)]

    # Combine
    X = [0] * N #array of zeros with size N
    for k in range(N // 2):#loop for half elements of k
        twiddle = np.exp(-2j * np.pi * k / N)#twiddle factor
        X[k] = even[k] + twiddle * odd[k]#first half of the elements of x
        X[k + N//2] = even[k] - twiddle * odd[k]#second half of the elements of x
    return X
```

```python
import matplotlib.pyplot as plt

# Generate a signal of length N=8 (must be a power of 2)
N = 16384
x = np.random.randint(-10, 10, size=N)+1j*np.random.randint(-10, 10, size=N).astype(complex)
p=0
for i in range(N):
    p+=x[i]

# Perform DIT-FFT
X = dit_fft(x)

# Print input and output
print(abs(p))
print("Input x:", x)
print("FFT X:", X)
```
